# Part A: Linear Regression

```python
from sklearn.linear_model import LinearRegression
from preprocessing import load_and_preprocess
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, r2_score
import matplotlib.pyplot as plt
from sklearn.pipeline import Pipeline
import joblib
```

## Load data and build the pipeline

```python
X_train, X_val, X_test, y_train, y_val, y_test, preprocessing = load_and_preprocess()

lg_pipelined = Pipeline([
    ("preprocessing", preprocessing),
    ("lg", LinearRegression())])
```

## Train and predict

```python
lg_pipelined.fit(X_train, y_train)
y_pred = lg_pipelined.predict(X_test)
```

## Metrics on the test set

```python
print(f"Root Mean Squared Error {root_mean_squared_error(y_test, y_pred)}")
print(f"Mean Absolute Error {mean_absolute_error(y_test, y_pred)}")
print(f"R Squared Score {r2_score(y_test, y_pred)}")
```

## Predicted vs Actual

```python
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred, alpha=0.5)
# perfect prediction line >>> pred = actual
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    "r--")
plt.xlabel("Actual values")
plt.ylabel("Predicted values")
plt.title("Predicted vs Actual")
plt.tight_layout()
plt.show()
```

## Save the model

```python
joblib.dump(lg_pipelined, '../models/LinearReg.pkl')
```
